```python
import numpy as np
import healpy as hp
from matplotlib import pyplot as plt
import pysm3
import pysm3.units as u
import pymaster as nmt
%matplotlib inline
#import packages
```

Add a galaxy cut to get rid of emissions from the galactic plane. Create Mask

```python
def galactic_strip_mask(nside):
    """Creates a HEALPix mask that smoothly transitions along the Galactic plane."""
    
    npix = hp.nside2npix(nside)  # Number of HEALPix pixels
    theta, phi = hp.pix2ang(nside, np.arange(npix))  # Get theta, phi for all pixels
    
    # Convert colatitude to Galactic latitude b
    #HEALPix colatitude 𝜃 to Galactic latitude b

    b = 90 - np.degrees(theta)
    bmin= 4
    bmax=10

    # Initialize mask
    mask = np.zeros(npix)

    # Apply cosine smoothing in the range 8° < |b| < 10°
    transition = (np.abs(b) > bmin) & (np.abs(b) < bmax)
    mask[transition] = 0.5 * (1 - np.cos(np.pi * (np.abs(b[transition]) - bmin) / (bmax-bmin)))

    # Set mask to 1 for |b| ≥ 10°
    mask[np.abs(b) >= 10] = 1

    return mask

# Set HEALPix resolution
nside = 128  # Adjust resolution for finer details

# Generate and plot the mask
mask = galactic_strip_mask(nside)
hp.mollview(mask, title="Galactic Plane Cosine Smoothed Mask")

plt.show()
```

```python
b_values = np.linspace(8, 10, 100)  # Simulate |b| from 8° to 10°
transition_function = 0.5 * (1 - np.cos(np.pi * (b_values - 8) / 2))

plt.plot(b_values, transition_function, label='Cosine Transition')
plt.xlabel('|b| (degrees)')
plt.ylabel('Mask Value')
plt.title('Smooth Transition from 0 to 1')
plt.legend()
plt.grid()
plt.show()
```
